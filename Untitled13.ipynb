```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.mixture import BayesianGaussianMixture
from sklearn.preprocessing import StandardScaler
from matplotlib.patches import Ellipse

# Define the dataset (Simulated 10 clusters with probabilities)
clusters = {
    "K1": (["Hypertension (0.30)", "Heart Disease (0.25)", "Obesity (0.20)"],
           ["Age 60-69 (0.40)", "BMI Obese (0.30)", "Family History (0.20)", "SES Low (0.10)"]),
    "K2": (["Diabetes (0.35)", "Kidney Disease (0.25)", "Neuropathy (0.15)"],
           ["Age 50-59 (0.35)", "Family History of Diabetes (0.30)", "SES Low (0.25)", "BMI Overweight (0.10)"]),
    "K3": (["COPD (0.40)", "Asthma (0.30)", "Anxiety (0.10)"],
           ["Male (0.35)", "Age 70+ (0.30)", "Smoking (0.25)", "Urban Residence (0.10)"]),
    "K4": (["Stroke (0.30)", "Dyslipidemia (0.25)", "Osteoporosis (0.20)"],
           ["Female (0.40)", "Age 60-69 (0.30)", "BMI Overweight (0.30)"]),
    "K5": (["Cancer (0.30)", "Hypertension (0.25)", "Obesity (0.15)"],
           ["Age 70+ (0.40)", "SES High (0.30)", "Family History (0.30)"]),
    "K6": (["Retinopathy (0.35)", "Obesity (0.30)", "Kidney Disease (0.20)"],
           ["Age 50-59 (0.35)", "BMI Obese (0.30)", "Family History of Diabetes (0.25)", "Smoking (0.10)"]),
    "K7": (["Cardiac Dysrhythmia (0.40)", "Stroke (0.30)", "Anxiety (0.15)"],
           ["Age 60-69 (0.35)", "SES Low (0.30)", "Smoking (0.25)", "Male (0.10)"]),
    "K8": (["COPD (0.45)", "Asthma (0.35)", "Hypertension (0.15)"],
           ["Male (0.40)", "Smoking (0.30)", "Age 70+ (0.20)", "Urban Residence (0.10)"]),
    "K9": (["Dyslipidemia (0.40)", "Heart Disease (0.25)", "Osteoporosis (0.20)"],
           ["Age 60-69 (0.35)", "BMI Overweight (0.30)", "Female (0.35)"]),
    "K10": (["Kidney Disease (0.35)", "Cancer (0.25)", "Diabetes (0.20)"],
            ["Age 70+ (0.40)", "Family History (0.30)", "SES High (0.30)"])
}

# Simulate data for 10 clusters (100 points per cluster)
np.random.seed(42)
n_clusters = len(clusters)
n_samples = 100 * n_clusters  # Total points

# Create synthetic 5D data
X = np.random.rand(n_samples, 5)

# Standardize data before PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Assign cluster labels
labels = np.repeat(list(clusters.keys()), 100)

# Apply PCA to reduce to 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Fit Bayesian Gaussian Mixture Model (DPMM)
dpmm = BayesianGaussianMixture(n_components=n_clusters, weight_concentration_prior=1e-2, random_state=42)
dpmm.fit(X_pca)
cluster_assignments = dpmm.predict(X_pca)

# Visualization setup
plt.figure(figsize=(12, 8))
sns.set(style="whitegrid")

colors = sns.color_palette("tab10", n_clusters)
markers = ['o', 's', 'D', 'v', '^', 'P', '*', 'X', 'H', '<']  # Different markers

# Scatter plot with distinct markers and colors
for i in range(n_clusters):
    cluster_points = X_pca[cluster_assignments == i]
    plt.scatter(cluster_points[:, 0], cluster_points[:, 1],
                s=50, color=colors[i], marker=markers[i], label=f"{list(clusters.keys())[i]}")

    # Fit an ellipse to each cluster
    cov = np.cov(cluster_points.T)
    v, w = np.linalg.eigh(cov)
    v = 2.8 * np.sqrt(v)  # Scale factor for better visualization
    angle = np.degrees(np.arctan2(w[1, 0], w[0, 0]))

    ellipse = Ellipse(xy=cluster_points.mean(axis=0), width=v[0], height=v[1],
                      angle=angle, edgecolor=colors[i], facecolor=colors[i], alpha=0.2, linestyle="--")
    plt.gca().add_patch(ellipse)

    # Improve labeling with arrows
    label_x, label_y = cluster_points.mean(axis=0)
    diseases, risks = clusters[list(clusters.keys())[i]]
    label_text = "\n".join(diseases + risks)

    plt.annotate(label_text, xy=(label_x, label_y), xytext=(label_x + 0.5, label_y + 0.5),
                 fontsize=9, weight='bold', bbox=dict(facecolor='white', alpha=0.6),
                 arrowprops=dict(arrowstyle="->", color=colors[i]))

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Enhanced Visualization of NCD Clusters using DPMM")
plt.legend(loc="upper right", fontsize=9, markerscale=1.2)
plt.show()
```

```python
# Step 1: Install pgmpy (Run this first in Google Colab)
!pip install pgmpy

# Step 2: Import required libraries
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from pgmpy.models import BayesianModel
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Step 3: Define the Bayesian Network structure
bn = BayesianModel([
    # Cluster K1
    ("Age_60_69", "Hypertension"), ("BMI_Obese", "Hypertension"), ("Family_History", "Hypertension"), ("SES_Low", "Hypertension"),
    ("Age_60_69", "Heart_Disease"), ("BMI_Obese", "Heart_Disease"), ("Family_History", "Heart_Disease"), ("SES_Low", "Heart_Disease"),
    ("Age_60_69", "Obesity"), ("BMI_Obese", "Obesity"), ("Family_History", "Obesity"), ("SES_Low", "Obesity"),

    # Cluster K2
    ("Age_50_59", "Diabetes"), ("Family_History_Diabetes", "Diabetes"), ("SES_Low", "Diabetes"), ("BMI_Overweight", "Diabetes"),
    ("Age_50_59", "Kidney_Disease"), ("Family_History_Diabetes", "Kidney_Disease"), ("SES_Low", "Kidney_Disease"), ("BMI_Overweight", "Kidney_Disease"),
    ("Age_50_59", "Neuropathy"), ("Family_History_Diabetes", "Neuropathy"), ("SES_Low", "Neuropathy"), ("BMI_Overweight", "Neuropathy"),

    # Cluster K3
    ("Male", "COPD"), ("Age_70_Plus", "COPD"), ("Smoking", "COPD"), ("Urban_Residence", "COPD"),
    ("Male", "Asthma"), ("Age_70_Plus", "Asthma"), ("Smoking", "Asthma"), ("Urban_Residence", "Asthma"),
    ("Male", "Anxiety"), ("Age_70_Plus", "Anxiety"), ("Smoking", "Anxiety"), ("Urban_Residence", "Anxiety"),

    # Multimorbidity node
    ("Hypertension", "Multimorbidity"), ("Heart_Disease", "Multimorbidity"), ("Obesity", "Multimorbidity"),
    ("Diabetes", "Multimorbidity"), ("Kidney_Disease", "Multimorbidity"), ("Neuropathy", "Multimorbidity"),
    ("COPD", "Multimorbidity"), ("Asthma", "Multimorbidity"), ("Anxiety", "Multimorbidity")
])

# Step 4: Define Conditional Probability Tables (CPTs)
cpd_age_60_69 = TabularCPD(variable="Age_60_69", variable_card=2, values=[[0.6], [0.4]])  # 40% have Age 60-69

cpd_hypertension = TabularCPD(variable="Hypertension", variable_card=2,
                              values=[[0.7, 0.6, 0.8, 0.75],  # P(No Hypertension)
                                      [0.3, 0.4, 0.2, 0.25]], # P(Yes Hypertension)
                              evidence=["Age_60_69", "BMI_Obese"],
                              evidence_card=[2, 2])

cpd_diabetes = TabularCPD(variable="Diabetes", variable_card=2,
                          values=[[0.65, 0.55, 0.75, 0.7],  # P(No Diabetes)
                                  [0.35, 0.45, 0.25, 0.3]],  # P(Yes Diabetes)
                          evidence=["Age_50_59", "Family_History_Diabetes"],
                          evidence_card=[2, 2])

cpd_multimorbidity = TabularCPD(variable="Multimorbidity", variable_card=2,
                                values=[[0.5, 0.6, 0.7, 0.65],  # P(No Multimorbidity)
                                        [0.5, 0.4, 0.3, 0.35]], # P(Yes Multimorbidity)
                                evidence=["Hypertension", "Diabetes"],
                                evidence_card=[2, 2])

# Step 5: Add CPDs to the model
bn.add_cpds(cpd_age_60_69, cpd_hypertension, cpd_diabetes, cpd_multimorbidity)

# Step 6: Verify model validity
assert bn.check_model()

# Step 7: Perform inference using Variable Elimination
inference = VariableElimination(bn)
prob_multimorbidity = inference.query(variables=["Multimorbidity"], evidence={"Age_60_69": 1, "BMI_Obese": 1})
print(prob_multimorbidity)

# Step 8: Draw the DAG with labeled edges
plt.figure(figsize=(12, 8))
G = nx.DiGraph()
G.add_edges_from(bn.edges())

pos = nx.spring_layout(G, seed=42)  # Layout for better visualization
nx.draw(G, pos, with_labels=True, node_size=3000, node_color="lightblue", edge_color="gray", font_size=10, font_weight="bold")

# Label edges with CPT values
edge_labels = {
    ("Age_60_69", "Hypertension"): "P=0.40",
    ("Age_50_59", "Diabetes"): "P=0.35",
    ("Hypertension", "Multimorbidity"): "P=0.30",
    ("Diabetes", "Multimorbidity"): "P=0.35"
}

nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color="red", font_size=10)

plt.title("Bayesian Network with Conditional Probabilities")
plt.show()
```

```python
from pgmpy.models import BayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Define the structure of the Bayesian Network with feedback loops
model = BayesianNetwork([
    ("Age 60-69", "Hypertension"), ("BMI Obese", "Heart Disease"),
    ("Family History", "Hypertension"), ("SES Low", "Hypertension"),
    ("Hypertension", "Obesity"), ("Obesity", "Heart Disease"),
    ("Heart Disease", "K9"), ("Obesity", "K5"), ("SES Low", "K2"),

    ("Age 50-59", "Obesity"), ("Family History of Diabetes", "Diabetes"),
    ("SES Low", "Diabetes"), ("BMI Overweight", "Obesity"),
    ("Diabetes", "K6"), ("Kidney Disease", "Hypertension"),
    ("Neuropathy", "K10"), ("Overweight", "K1"),

    ("Male", "COPD"), ("Age 70+", "COPD"), ("Smoking", "COPD"),
    ("Urban Residence", "COPD"), ("COPD", "Smoking"), ("COPD", "Urban Residence"),
    ("Asthma", "K8"), ("Anxiety", "K7"), ("Urban Residence", "K8"),

    ("Female", "Stroke"), ("Age 60-69", "Stroke"), ("BMI Overweight", "Stroke"),
    ("Stroke", "Dyslipidemia"), ("Dyslipidemia", "K9"),
    ("Osteoporosis", "K9"),

    ("Age 70+", "Cancer"), ("SES High", "Cancer"), ("Family History", "Cancer"),
    ("Cancer", "Hypertension"), ("Hypertension", "K1"), ("Obesity", "K1"),

    ("Age 50-59", "Retinopathy"), ("BMI Obese", "Retinopathy"),
    ("Family History of Diabetes", "Kidney Disease"),
    ("Kidney Disease", "K10"), ("Smoking", "K3"),

    ("Age 60-69", "Cardiac Dysrhythmia"), ("SES Low", "Cardiac Issues"),
    ("Smoking", "Cardiac Issues"), ("Male", "K3"),

    ("Male", "COPD"), ("Smoking", "COPD"), ("Age 70+", "COPD"),
    ("COPD", "Hypertension"), ("Urban Residence", "K3"),

    ("Age 60-69", "Dyslipidemia"), ("BMI Overweight", "Dyslipidemia"),
    ("Female", "Dyslipidemia"),

    ("Age 70+", "Kidney Disease"), ("Family History", "Kidney Disease"),
    ("SES High", "Kidney Disease"), ("Diabetes", "K2"),
])

# Example of defining CPDs (conditional probability distributions)
cpd_age_60_69 = TabularCPD(variable='Age 60-69', variable_card=2, values=[[0.6], [0.4]])
cpd_hypertension = TabularCPD(variable='Hypertension', variable_card=2,
                               values=[[0.7, 0.6], [0.3, 0.4]],
                               evidence=['Age 60-69'], evidence_card=[2])

# Add CPDs to the model
model.add_cpds(cpd_age_60_69, cpd_hypertension)

# Check model validity
assert model.check_model()

# Perform inference
inference = VariableElimination(model)
result = inference.map_query(variables=['Hypertension'])
print("Most probable state of Hypertension:", result)
```

```python
import networkx as nx
!pip install pygraphviz
import pygraphviz as pgv
from networkx.drawing.nx_agraph import to_agraph
import matplotlib.pyplot as plt

# Create a directed graph
G = nx.DiGraph()

# Define clusters with their associated probabilities
clusters = {
    "K1": ("Hypertension, Heart Disease, Obesity", 0.21, 0.79),
    "K2": ("Diabetes, Kidney Disease, Neuropathy", 0.30, 0.70),
    "K3": ("COPD, Asthma, Anxiety", 0.18, 0.82),
    "K4": ("Stroke, Dyslipidemia, Osteoporosis", 0.35, 0.65),
    "K5": ("Cancer, Hypertension, Obesity", 0.25, 0.75),
    "K6": ("Retinopathy, Diabetes, Kidney Disease", 0.27, 0.73),
    "K7": ("Cardiac Dysrhythmia, Stroke, Anxiety", 0.22, 0.78),
    "K8": ("COPD, Asthma, Hypertension", 0.20, 0.80),
    "K9": ("Dyslipidemia, Heart Disease, Osteoporosis", 0.33, 0.67),
    "K10": ("Kidney Disease, Cancer, Diabetes", 0.28, 0.72)
}

# Define key risk factors associated with each cluster
cluster_risk_factors = {
    "K1": ["Age 60-69", "BMI Obese", "Family History", "SES Low"],
    "K2": ["Age 50-59", "Family History of Diabetes", "SES Low", "BMI Overweight"],
    "K3": ["Male", "Age 70+", "Smoking", "Urban Residence"],
    "K4": ["Female", "Age 60-69", "BMI Overweight"],
    "K5": ["Age 70+", "SES High", "Family History"],
    "K6": ["Age 50-59", "BMI Obese", "Family History of Diabetes", "Smoking"],
    "K7": ["Age 60-69", "SES Low", "Smoking", "Male"],
    "K8": ["Male", "Smoking", "Age 70+", "Urban Residence"],
    "K9": ["Age 60-69", "BMI Overweight", "Female"],
    "K10": ["Age 70+", "Family History", "SES High"]
}

# Add edges from risk factors to clusters
for cluster, factors in cluster_risk_factors.items():
    for factor in factors:
        G.add_edge(factor, cluster)

# Add edges from clusters to dominant NCDs
for cluster, (ncds, p_no, p_yes) in clusters.items():
    for ncd in ncds.split(", "):
        G.add_edge(cluster, ncd)

# Add edges from NCDs to multimorbidity outcome
for cluster in clusters.keys():
    G.add_edge(cluster, "Multimorbidity")

# Assign labels with probabilities
node_labels = {}
for cluster, (ncds, p_no, p_yes) in clusters.items():
    node_labels[cluster] = f"{cluster}\nP(No): {p_no}\nP(Yes): {p_yes}"

# Draw the DAG
plt.figure(figsize=(15, 10))
A = to_agraph(G)
for node, label in node_labels.items():
    A.get_node(node).attr["label"] = label  # Assign labels to nodes
A.layout(prog="dot")  # Use Graphviz layout
A.draw("dag.png")  # Save the DAG as an image

# Show the graph
img = plt.imread("dag.png")
plt.imshow(img)
plt.axis("off")
plt.show()
```
